```python
# 2 Implementation of Document Retrieval Using Inverted Files
```

```python
# Collection of documents
documents = {
    1: "Information retrieval is used to search documents",
    2: "Search engines retrieve relevant information",
    3: "Information retrieval systems help users find documents",
    4: "Python is useful for information processing"
}
```

```python
# -----------------------------------------
# Create Inverted Index
# -----------------------------------------

inverted_index = {}

for doc_id, text in documents.items():
    # Convert text to lowercase
    words = text.lower().split()

    for word in words:
        # Remove punctuation
        word = word.strip(".,!?")

        # Add document ID to inverted index
        if word not in inverted_index:
            inverted_index[word] = []

        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)
```

```python
# -----------------------------------------
# Display Inverted Index
# -----------------------------------------

print("INVERTED INDEX")
print("-------------------------")

for word in sorted(inverted_index):
    print(word, ":", inverted_index[word])
```

```python
# -----------------------------------------
# Document Retrieval
# -----------------------------------------

query = input("\nEnter search term: ").lower()

if query in inverted_index:
    print("\nRetrieved Documents:")

    for doc_id in inverted_index[query]:
        print("Document", doc_id, ":", documents[doc_id])
else:
    print("\nNo documents found for the given query.")
```
