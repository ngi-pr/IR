```python
# Program to Calculate Precision and Recall

# Answer set A - Retrieved documents
A = {"D1", "D2", "D3", "D4"}

# Relevant documents for query q1
Rq1 = {"D1", "D3", "D5", "D6"}

# Find relevant retrieved documents
relevant_retrieved = A.intersection(Rq1)

# Calculate Precision
precision = len(relevant_retrieved) / len(A)

# Calculate Recall
recall = len(relevant_retrieved) / len(Rq1)

# Display results
print("Answer Set A:", A)
print("Relevant Documents Rq1:", Rq1)
print("Relevant Retrieved Documents:", relevant_retrieved)
print("Precision =", precision)
print("Recall =", recall)
```
