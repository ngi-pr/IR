```python
import cv2
import matplotlib.pyplot as plt

# Read the input image (upload 1.webp to /content/ first)
image = cv2.imread("param.jpg")

if image is None:
    raise FileNotFoundError("param.jpg")

# Convert BGR image to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Separate color channels
red = image[:, :, 0]
green = image[:, :, 1]
blue = image[:, :, 2]

# Display original image and RGB channel histograms
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.hist(red.ravel(), bins=256, range=(0, 256))
plt.title("Red Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(green.ravel(), bins=256, range=(0, 256))
plt.title("Green Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(blue.ravel(), bins=256, range=(0, 256))
plt.title("Blue Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
